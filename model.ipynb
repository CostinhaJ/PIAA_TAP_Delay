```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import time
import re
import random
```

```python
#############################################################
# Carregar e Explorar
#############################################################
def load_and_explore_data(file_path: str, nrows=582000):
    data = pd.read_csv(file_path, delimiter=',')#, nrows=nrows)

    return data

file_path = '/flight_data_2018_2024.csv'
data_original = load_and_explore_data(file_path)

print(data_original.shape)
print(data_original.head(1))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Convert FlightDate column to datetime format
data_original['FlightDate'] = pd.to_datetime(data_original['FlightDate'])

# Define the airport code of interest (change this as needed)
airport_code = 'ATL'

# Filter for departures (origin) and arrivals (destination)
departures = data_original[data_original['Origin'] == airport_code]
arrivals   = data_original[data_original['Dest'] == airport_code]

# Group by FlightDate and count the number of flights
dep_counts = departures.groupby('FlightDate').size()
arr_counts = arrivals.groupby('FlightDate').size()

# Combine counts into one DataFrame
flight_counts = pd.DataFrame({
    'Departures': dep_counts,
    'Arrivals': arr_counts
}).fillna(0)

# Create a full date range from 2018-01-01 to 2024-12-31
full_date_range = pd.date_range(start='2018-01-01', end='2024-12-31')

# Reindex the flight_counts DataFrame to include every date in the full range, filling missing dates with 0
flight_counts = flight_counts.reindex(full_date_range, fill_value=0)

# Plot the data
plt.figure(figsize=(12, 6))
plt.plot(flight_counts.index, flight_counts['Departures'], label='Departures', marker='o')
plt.plot(flight_counts.index, flight_counts['Arrivals'], label='Arrivals', marker='o')
plt.xlabel('Date')
plt.ylabel('Number of Flights')
plt.title(f'Incoming vs Departure Flights by Date for {airport_code}')
plt.legend()
plt.xticks(rotation=45)
# Set the x-axis limits explicitly to the full date range
plt.xlim(pd.Timestamp("2018-01-01"), pd.Timestamp("2024-12-31"))
plt.tight_layout()
plt.show()
```
